```python
k = 100
plt.plot(np.arange(k), S[:k])
plt.title('Сингулярні значення')
plt.xlabel('k')
plt.ylabel('Значення')
plt.show()
```

```python

for k in [5, 20, 50, 100]:
    svd = TruncatedSVD(n_components=k)
    truncated = svd.fit_transform(flat_image2)
    reconstructed = svd.inverse_transform(truncated)
    #рахуємо помилку
    mse = np.mean(np.square(reconstructed - flat_image2))

    height, width, channels = image2.shape
    reconstructed = reconstructed.reshape(height, width, channels)
    reconstructed = np.clip(reconstructed, 0, 1)  # діапазон 0-1!

    plt.imshow(reconstructed)
    plt.title(f'k={k}, MSE={mse:.4f}')
    plt.axis('off')
    plt.show()
```

Висновки
Логотип:
k=5  → 0.0244
k=50 → 0.0024

Пейзаж ч/б:
k=5  → 0.0062
k=50 → 0.0003  ← краще ніж логотип!

Природа колір (зменшена):
k=5  → 0.0071
k=50 → 0.0008

1. SVD ефективніше стискає прості зображення (логотипи)
   ніж складні фотографії.

2. Чорно-біле зображення стискається краще ніж кольорове
   бо має 1 канал замість 3.

3. Для логотипу достатньо k=50, для пейзажу k=100,
   а кольорова природа потребує k=500+.

4. Чим швидше спадають сингулярні значення —
   тим краще стискається зображення.

5. пейзаж ч/б краще стискається ніж логотип,бо плавні лінії лешще описати математично.
