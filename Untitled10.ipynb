```python
import pandas as pd

df = pd.read_csv('/content/2017.csv')
print(df.shape)
print(df.columns.tolist())
df.head()
```

```python
import matplotlib.pyplot as plt

features = ['Economy..GDP.per.Capita.', 'Family',
            'Health..Life.Expectancy.', 'Freedom',
            'Generosity', 'Trust..Government.Corruption.']

df[features].hist(bins=20, figsize=(12, 8))
plt.tight_layout()
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler

features = ['Economy..GDP.per.Capita.', 'Family',
            'Health..Life.Expectancy.', 'Freedom']

X = df[features].dropna()
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Оригінал - середнє:", X.mean().round(3).tolist())
print("Після стандартизації - середнє:", X_scaled.mean(axis=0).round(3).tolist())
```

```python
from sklearn.mixture import GaussianMixture
import matplotlib.pyplot as plt

bic_scores = []
k_range = range(2, 10)

for k in k_range:
    gmm = GaussianMixture(n_components=k, random_state=42)
    gmm.fit(X_scaled)
    bic_scores.append(gmm.bic(X_scaled))

plt.plot(k_range, bic_scores, 'bo-')
plt.xlabel('Кількість кластерів')
plt.ylabel('BIC')
plt.title('Вибір кількості кластерів')
plt.show()
```

```python
gmm = GaussianMixture(n_components=3, random_state=42)
gmm.fit(X_scaled)
df['Cluster'] = gmm.predict(X_scaled)
print(df['Cluster'].value_counts())
```

```python
print(df.groupby('Cluster')['Happiness.Score'].mean().round(2))
```

```python
print(df[df['Cluster']==2]['Country'].tolist())
```

```python
import plotly.express as px

fig = px.choropleth(df,
                    locations="Country",
                    color="Cluster",
                    locationmode="country names",
                    color_discrete_sequence=["#FF6B6B", "#4ECDC4", "#45B7D1"],
                    title="Кластери щастя 2017")
fig.show()
```

```python
fig = px.choropleth(df,
                    locations="Country",
                    color=df['Cluster'].astype(str),
                    locationmode="country names",
                    color_discrete_map={"0": "#4ECDC4", "1": "#FF6B6B", "2": "#45B7D1"},
                    title="Кластери щастя 2017")
fig.show()
```

 Висновки

 1. Аналіз розподілу ознак
Generosity та Trust..Government.Corruption. мають сильний правосторонній
нахил — виключені з моделі. Для GMM обрано 4 ознаки: GDP, Family,
Health, Freedom.

 2. Стандартизація
StandardScaler привів всі ознаки до середнього=0, дисперсія=1.
Без стандартизації GDP домінував би над іншими ознаками.

 3. Вибір кількості кластерів (BIC)
BIC мінімальний при k=3 — математично підтверджує логічний поділ
країн на 3 групи.

4. Результати кластеризації

| Кластер | Країн | Середнє щастя | Регіон |
|---------|-------|---------------|--------|
| 2 | 34 | 6.68 | Європа, Скандинавія |
| 0 | 77 | 5.50 | Америка, Азія |
| 1 | 44 | 4.06 | Африка |

 5. Інтерпретація
Африканські країни (кластер 1) мають низький рівень GDP,
здоров'я, сімейної підтримки та свободи.
Скандинавські країни (кластер 2) лідирують за всіма показниками.

```python
features2 = ['Economy..GDP.per.Capita.', 'Health..Life.Expectancy.']

X2 = df[features2].dropna()
X2_scaled = scaler.fit_transform(X2)

gmm2 = GaussianMixture(n_components=3, random_state=42)
gmm2.fit(X2_scaled)
df['Cluster2'] = gmm2.predict(X2_scaled)

print(df.groupby('Cluster2')['Happiness.Score'].mean().round(2))
```

```python
changed = df[df['Cluster'] != df['Cluster2']][['Country', 'Cluster', 'Cluster2']]
print(changed)
```

```python
dfs = {}
for year in [2015, 2016, 2017, 2018, 2019]:
    dfs[year] = pd.read_csv(f'/content/{year}.csv')
    print(f"{year}: {dfs[year].shape} | колонки: {dfs[year].columns.tolist()}")
```

```python
rename_map = {
    2015: {'Country': 'Country', 'Happiness Score': 'Score',
           'Economy (GDP per Capita)': 'GDP', 'Family': 'Family',
           'Health (Life Expectancy)': 'Health', 'Freedom': 'Freedom'},
    2016: {'Country': 'Country', 'Happiness Score': 'Score',
           'Economy (GDP per Capita)': 'GDP', 'Family': 'Family',
           'Health (Life Expectancy)': 'Health', 'Freedom': 'Freedom'},
    2017: {'Country': 'Country', 'Happiness.Score': 'Score',
           'Economy..GDP.per.Capita.': 'GDP', 'Family': 'Family',
           'Health..Life.Expectancy.': 'Health', 'Freedom': 'Freedom'},
    2018: {'Country or region': 'Country', 'Score': 'Score',
           'GDP per capita': 'GDP', 'Social support': 'Family',
           'Healthy life expectancy': 'Health',
           'Freedom to make life choices': 'Freedom'},
    2019: {'Country or region': 'Country', 'Score': 'Score',
           'GDP per capita': 'GDP', 'Social support': 'Family',
           'Healthy life expectancy': 'Health',
           'Freedom to make life choices': 'Freedom'}
}
```

```python
all_dfs = []
for year in [2015, 2016, 2017, 2018, 2019]:
    df_year = dfs[year].rename(columns=rename_map[year])
    df_year = df_year[['Country', 'Score', 'GDP', 'Family', 'Health', 'Freedom']]
    df_year['Year'] = year
    all_dfs.append(df_year)

df_all = pd.concat(all_dfs, ignore_index=True)
print(df_all.shape)
```

```python
features = ['GDP', 'Family', 'Health', 'Freedom']

X_all = df_all[features].dropna()
scaler2 = StandardScaler()
X_all_scaled = scaler2.fit_transform(X_all)

gmm_all = GaussianMixture(n_components=3, random_state=42)
gmm_all.fit(X_all_scaled)
df_all['Cluster'] = gmm_all.predict(X_all_scaled)

print(df_all.groupby('Cluster')['Score'].mean().round(2))
```

```python
print(df_all.groupby(['Year', 'Cluster']).size().unstack())
```

```python
fig = px.choropleth(df_all[df_all['Year']==2019],
                    locations="Country",
                    color=df_all[df_all['Year']==2019]['Cluster'].astype(str),
                    locationmode="country names",
                    color_discrete_map={"0": "#FF6B6B", "1": "#45B7D1", "2": "#4ECDC4"},
                    title="Кластери щастя 2019")
fig.show()
```

```python
import matplotlib.pyplot as plt

cluster_by_year = df_all.groupby(['Year', 'Cluster']).size().unstack()

cluster_by_year.plot(kind='bar', figsize=(10, 6),
                     color=['#FF6B6B', '#45B7D1', '#4ECDC4'])
plt.xlabel('Рік')
plt.ylabel('Кількість країн')
plt.title('Розподіл кластерів по роках')
plt.legend(['Нещасливі', 'Щасливі', 'Середні'])
plt.tight_layout()
plt.show()
```
