```python
def compute_value_function(policy, env, gamma=0.9, theta=1e-10):
    V = np.zeros(env.observation_space.n)  # початкові значення
    while True:
        delta = 0
        for s in range(env.observation_space.n):
            v = 0
            action = policy[s]
            for prob, next_state, reward, done in env.unwrapped.P[s][action]:
                v += prob * (reward + gamma * V[next_state] * (1 - done))
            delta = max(delta, abs(v - V[s]))
            V[s] = v
        if delta < theta:  # критерій збіжності
            break
    return V
```

```python
def policy_iteration(env, gamma=0.9, theta=1e-10):
    # Початкова випадкова політика
    policy = np.zeros(env.observation_space.n, dtype=int)

    while True:
        # Крок 1: оцінюємо поточну політику
        V = compute_value_function(policy, env, gamma, theta)

        # Крок 2: покращуємо політику
        policy_stable = True

        for s in range(env.observation_space.n):
            old_action = policy[s]

            # Рахуємо цінність кожної дії
            action_values = np.zeros(env.action_space.n)
            for a in range(env.action_space.n):
                for prob, next_state, reward, done in env.unwrapped.P[s][a]:
                    action_values[a] += prob * (reward + gamma * V[next_state] * (1 - done))

            # Обираємо найкращу дію
            policy[s] = np.argmax(action_values)

            # Чи змінилась політика?
            if old_action != policy[s]:
                policy_stable = False

        # Крок 3: якщо політика стабільна - стоп
        if policy_stable:
            return policy, V
```

```python
policy, V = policy_iteration(env)
print("Політика:", policy)
print("Цінності:", V)
```

```python
def show_policy(policy):
    arrows = {0: '←', 1: '↓', 2: '→', 3: '↑'}
    grid = ['H' if V[i] == 0 and i != 0 else arrows[policy[i]] for i in range(16)]
    grid[15] = 'G'
    for i in range(0, 16, 4):
        print(' '.join(grid[i:i+4]))

show_policy(policy)
```

```python

def show_policy_visual(policy, V):
    arrows = {0: '←', 1: '↓', 2: '→', 3: '↑'}
    desc = ['SFFF', 'FHFH', 'FFFH', 'HFFG']

    fig, ax = plt.subplots(figsize=(6, 6))

    for i in range(4):
        for j in range(4):
            s = i * 4 + j
            cell = desc[i][j]

            # Колір клітинки
            if cell == 'H':
                color = '#2196F3'  # синій - ополонка
            elif cell == 'G':
                color = '#4CAF50'  # зелений - ціль
            elif s == 0:
                color = '#FF9800'  # помаранчевий - старт
            else:
                color = '#F5F5F5'  # сірий - лід

            rect = plt.Rectangle([j, 3-i], 1, 1,
                               facecolor=color, edgecolor='black', linewidth=2)
            ax.add_patch(rect)

            # Стрілка або символ
            if cell == 'H':
                ax.text(j+0.5, 3-i+0.5, '💀',
                       ha='center', va='center', fontsize=20)
            elif cell == 'G':
                ax.text(j+0.5, 3-i+0.5, '🏆',
                       ha='center', va='center', fontsize=20)
            elif s == 0:
                ax.text(j+0.5, 3-i+0.5, arrows[policy[s]],
                       ha='center', va='center', fontsize=24,
                       fontweight='bold', color='white')
            else:
                ax.text(j+0.5, 3-i+0.5, arrows[policy[s]],
                       ha='center', va='center', fontsize=24,
                       fontweight='bold', color='#333333')

            # Цінність стану
            ax.text(j+0.5, 3-i+0.15, f'{V[s]:.2f}',
                   ha='center', va='center', fontsize=9, color='#666666')

    ax.set_xlim(0, 4)
    ax.set_ylim(0, 4)
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_title('Оптимальна політика FrozenLake-v1\n(Policy Iteration)',
                fontsize=14, fontweight='bold', pad=15)

    # Легенда
    legend = [
        mpatches.Patch(color='#FF9800', label='Старт (S)'),
        mpatches.Patch(color='#2196F3', label='Ополонка (H)'),
        mpatches.Patch(color='#4CAF50', label='Ціль (G)'),
        mpatches.Patch(color='#F5F5F5', label='Лід (F)'),
    ]
    ax.legend(handles=legend, loc='upper center',
             bbox_to_anchor=(0.5, -0.05), ncol=4, fontsize=9)

    plt.tight_layout()
    plt.show()

show_policy_visual(policy, V)
```

```python
import time
from IPython.display import clear_output

def animate_agent(env, policy, episodes=3):
    arrows = {0: '←', 1: '↓', 2: '→', 3: '↑'}
    desc = ['SFFF', 'FHFH', 'FFFH', 'HFFG']

    for episode in range(episodes):
        state = env.reset()
        state = state[0] if isinstance(state, tuple) else state
        done = False
        step = 0

        print(f"\n=== Епізод {episode+1} ===")

        while not done and step < 50:
            clear_output(wait=True)
            print(f"=== Епізод {episode+1} | Крок {step} ===\n")

            for i in range(4):
                row = ''
                for j in range(4):
                    s = i * 4 + j
                    if s == state:
                        row += ' 🤖'
                    elif desc[i][j] == 'H':
                        row += ' 💀'
                    elif desc[i][j] == 'G':
                        row += ' 🏆'
                    else:
                        row += ' ' + arrows[policy[s]]
                print(row)

            action = policy[state]
            result = env.step(action)

            if len(result) == 5:
                next_state, reward, terminated, truncated, _ = result
                done = terminated or truncated
            else:
                next_state, reward, done, _ = result

            state = next_state
            step += 1
            time.sleep(0.5)

        if reward == 1:
            print("\n🎉 Досяг цілі!")
        else:
            print("\n💀 Впав в ополонку!")
        time.sleep(1)

animate_agent(env, policy)
```

Висновки

1. Policy Iteration
Алгоритм почергово оцінює поточну політику (compute_value_function)
та покращує її до стабілізації. Це класичний MDP на основі ланцюгів Маркова.

2. Стохастичність (is_slippery=True)
Агент не завжди іде туди куди хоче — імовірність 1/3 для кожного
з 3 напрямків. Тому оптимальна політика іноді виглядає "дивно" —
агент уникає ополонок навіть ціною обхідного маршруту.

3. Вплив gamma

| gamma | V(старт) | Поведінка |
|-------|----------|-----------|
| 0.5   | 0.0004   | Короткозорий — не бачить далекої цілі |
| 0.9   | 0.0689   | Збалансований — оптимальний вибір |
| 0.99  | 0.5420   | Далекозорий — максимально цінує майбутнє |
