```python
from sklearn.datasets import load_breast_cancer

cancer = load_breast_cancer()
print(cancer.keys())
```

```python
print(cancer.DESCR)
```

```python
import pandas as pd

df = pd.DataFrame(cancer.data, columns=cancer.feature_names)
df['target'] = cancer.target
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df.drop('target', axis=1))
print(X_scaled.shape)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

corr_matrix = df.drop('target', axis=1).corr()

plt.figure(figsize=(15, 12))
sns.heatmap(corr_matrix, cmap='coolwarm')
plt.title('Матриця кореляції')
plt.show()
```

```python
# Беремо тільки перші 5 ознак
cols = list(cancer.feature_names[:5]) + ['target']
sns.pairplot(df[cols], hue='target', palette='Set1')
plt.show()
```

```python
from sklearn.metrics import pairwise_distances

# Беремо перші 50 зразків щоб матриця була читабельна
X_subset = X_scaled[:50]

euclidean_dist = pairwise_distances(X_subset, metric='euclidean')
print(euclidean_dist.shape)
```

```python
metrics = ['euclidean', 'cityblock', 'cosine', 'l1', 'manhattan']

fig, axes = plt.subplots(1, 5, figsize=(25, 5))

for i, metric in enumerate(metrics):
    dist_matrix = pairwise_distances(X_subset, metric=metric)
    sns.heatmap(dist_matrix, ax=axes[i], cmap='viridis')
    axes[i].set_title(metric)

plt.show()
```

перші 50 зразків виявились злоякісними,тому це не є валідно



```python
# Сортуємо за target
df_sorted = df.sort_values('target')
X_sorted = X_scaled[df_sorted.index]
X_subset = X_sorted[:50]

fig, axes = plt.subplots(1, 5, figsize=(25, 5))

for i, metric in enumerate(metrics):
    dist_matrix = pairwise_distances(X_subset, metric=metric)
    sns.heatmap(dist_matrix, ax=axes[i], cmap='viridis')
    axes[i].set_title(metric)

plt.show()
```

```python
df_sorted = df.sort_values('target')
X_sorted = X_scaled[df_sorted.index]
X_subset = X_sorted[:50]
print(X_subset.shape)
print(df_sorted['target'][:50].unique())
```

тут знову вибрано тільки злоякісні,отже цей метод не є валідним

створюємо 2 класи по 25 зразки у кожному

```python
import numpy as np
malignant_idx = df_sorted[df_sorted['target'] == 0].index[:25]
benign_idx = df_sorted[df_sorted['target'] == 1].index[:25]

malignant_data = X_scaled[malignant_idx]
benign_data = X_scaled[benign_idx]

X_subset_mixed = np.vstack([malignant_data, benign_data])
print(X_subset_mixed.shape)
```

```python
metrics = ['euclidean', 'cityblock', 'cosine', 'l1', 'manhattan']

fig, axes = plt.subplots(1, 5, figsize=(25, 5))

for i, metric in enumerate(metrics):
    dist_matrix = pairwise_distances(X_subset_mixed, metric=metric)
    sns.heatmap(dist_matrix, ax=axes[i], cmap='viridis')
    axes[i].set_title(metric)

plt.show()
```

```python
df_mixed = pd.DataFrame(X_subset_mixed, columns=cancer.feature_names)
df_mixed['target'] = [0] * 25 + [1] * 25
print(df_mixed['target'].value_counts())
df_mixed.head()
```

```python
cols = list(cancer.feature_names[:5]) + ['target']
sns.pairplot(df_mixed[cols], hue='target', palette='Set1')
plt.show()
```
