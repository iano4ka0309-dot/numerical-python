```python
import numpy as np
import pygad


```

```python
initial_population = np.random.randint(low=0, high=2, size=(10, 14))
```

```python
def fitness_func(ga_instance, solution, solution_idx):
    price = np.sum(solution * values)
    weight_now = np.sum(solution * weights)
    if weight_now > max_weight:
        return 0
    else:
        return price
```

```python

ga_instance = pygad.GA(
    num_generations=100,
    num_parents_mating=5,
    fitness_func=fitness_func,
    initial_population=initial_population,
    crossover_type="two_points",
    mutation_type="random",
    gene_type=int
)
```

```python
class Product:
    def __init__(self, name, space, price):
        self.name = name
        self.space = space
        self.price = price
```

```python
products_list = []
products_list.append(Product('Refrigerator A', 0.751, 999.90))
products_list.append(Product('Cell phone', 0.00000899, 2199.12))
products_list.append(Product('TV 55', 0.400, 4346.99))
products_list.append(Product("TV 50' ", 0.290, 3999.90))
products_list.append(Product("TV 42' ", 0.200, 2999.00))
products_list.append(Product("Notebook A", 0.00350, 2499.90))
products_list.append(Product("Ventilator", 0.496, 199.90))
products_list.append(Product("Microwave A", 0.0424, 308.66))
products_list.append(Product("Microwave B", 0.0544, 429.90))
products_list.append(Product("Microwave C", 0.0319, 299.29))
products_list.append(Product("Refrigerator B", 0.635, 849.00))
products_list.append(Product("Refrigerator C", 0.870, 1199.89))
products_list.append(Product("Notebook B", 0.498, 1999.90))
products_list.append(Product("Notebook C", 0.527, 3999.00))
```

```python
values = np.array([p.price for p in products_list])
weights = np.array([p.space for p in products_list])
max_weight = 3.0
```

```python
ga_instance.run()
```

```python
solution, solution_fitness, _ = ga_instance.best_solution()
print("Найкраща хромосома:", solution)
print("Цінність:", solution_fitness)
```

```python
for i, gene in enumerate(solution):
    if gene == 1:
        print(products_list[i].name, products_list[i].price)
```

```python
total_weight = np.sum(solution * weights)
print("Загальна вага:", total_weight)
print("Ліміт:", max_weight)
```

```python

ga_instance = pygad.GA(
    num_generations=100,
    num_parents_mating=5,
    fitness_func=fitness_func,
    initial_population=initial_population,
    crossover_type="uniform",
    mutation_type="random",
    gene_type=int
)
ga_instance.run()
solution, solution_fitness, _ = ga_instance.best_solution()
print("Цінність:", solution_fitness)
```

```python
total_weight = np.sum(solution * weights)
print("Загальна вага:", total_weight)
print("Ліміт:", max_weight)
```

```python
ga_instance = pygad.GA(
    num_generations=100,
    num_parents_mating=5,
    fitness_func=fitness_func,
    initial_population=initial_population,
    crossover_type="single_point",
    mutation_type="swap",
    gene_type=int
)
ga_instance.run()
solution, solution_fitness, _ = ga_instance.best_solution()
print("Цінність:", solution_fitness)
```

```python
ga_instance = pygad.GA(
    num_generations=100,
    num_parents_mating=5,
    fitness_func=fitness_func,
    initial_population=initial_population,
    crossover_type="single_point",
    mutation_type="inversion",
    gene_type=int
)
ga_instance.run()
solution, solution_fitness, _ = ga_instance.best_solution()
print("Цінність:", solution_fitness)
```

```python
import matplotlib.pyplot as plt
labels = ["single+random", "two_points+random", "uniform+random", "uniform+swap", "uniform+inversion" ]
values_chart = [19635.18,19934.47,23431.38,20822.90,24281.55]
plt.bar(labels, values_chart)
plt.title("Порівняння кросоверів та мутацій")
plt.ylabel("Цінність")
plt.xticks(rotation=45, ha='right')
plt.show()
```

**Висновок **
найкращий результат дало uniform і inversion
